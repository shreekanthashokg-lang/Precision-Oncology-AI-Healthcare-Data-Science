# 02 - EDA: Genomics Data

Explore the (synthetic or ClinVar-derived) DNA variant dataset: class balance, sequence composition, and k-mer patterns.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from src.genomics.encoding import kmer_frequency_encode
```

```python
processed_dir = Path('../data/processed/genomics')
train_df = pd.read_csv(processed_dir / 'train.csv')
train_df.head()
```

## Class balance

```python
train_df['label_name'].value_counts().plot(kind='bar', title='Class distribution (train split)')
plt.ylabel('count')
plt.show()
```

## Base composition (GC content) by class

```python
def gc_content(seq):
    seq = seq.upper()
    return (seq.count('G') + seq.count('C')) / max(len(seq), 1)

train_df['gc_content'] = train_df['sequence'].apply(gc_content)
train_df.boxplot(column='gc_content', by='label_name')
plt.title('GC content by class')
plt.suptitle('')
plt.show()
```

## Average 4-mer frequency profile by class

```python
for label_name, group in train_df.groupby('label_name'):
    kmers = group['sequence'].apply(lambda s: kmer_frequency_encode(s, k=4))
    avg_profile = pd.DataFrame(kmers.tolist()).mean()
    print(label_name, '- top 5 4-mers:')
    print(avg_profile.sort_values(ascending=False).head())
```
