# 01 - EDA: Histopathology Data

Explore class balance, sample images, and basic statistics for the histopathology dataset.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import cv2
```

```python
processed_dir = Path('../data/processed/histopathology')
train_df = pd.read_csv(processed_dir / 'train.csv')
train_df.head()
```

## Class balance

```python
train_df['label'].value_counts().plot(kind='bar', title='Class distribution (train split)')
plt.ylabel('count')
plt.show()
```

## Sample images

```python
fig, axes = plt.subplots(2, 4, figsize=(14, 7))
sample = train_df.sample(min(8, len(train_df)), random_state=42).reset_index(drop=True)
for i, ax in enumerate(axes.flat):
    if i < len(sample):
        img = cv2.cvtColor(cv2.imread(sample.loc[i, 'path']), cv2.COLOR_BGR2RGB)
        ax.imshow(img)
        ax.set_title(sample.loc[i, 'label'])
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## Image size / basic stats

```python
sizes = [cv2.imread(p).shape[:2] for p in train_df['path'].sample(min(50, len(train_df)), random_state=42)]
pd.DataFrame(sizes, columns=['height', 'width']).describe()
```
