````python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

from src.utils.helpers import load_config
from src.utils.seed import set_seed

cfg = load_config('../config/config.yaml')
set_seed(cfg['seed'])
cfg['genomics']
````

## Quick sanity check: CNN vs Transformer forward pass

````python
import torch
from src.genomics.models import build_genomics_model

for model_type in ['cnn', 'transformer']:
    model = build_genomics_model(model_type, cfg['genomics']['seq_length'], cfg['genomics']['embedding_dim'], cfg['genomics']['num_classes'])
    dummy = torch.randn(2, 4, cfg['genomics']['seq_length'])
    logits = model(dummy)
    print(model_type, '-> logits shape:', logits.shape, '| embedding dim:', model.embed_dim)
````

## Attention visualization sanity check (Transformer only)

````python
from src.genomics.models import DNATransformerEncoder

transformer = DNATransformerEncoder(seq_length=cfg['genomics']['seq_length'], embedding_dim=cfg['genomics']['embedding_dim'])
dummy = torch.randn(1, 4, cfg['genomics']['seq_length'])
attn = transformer.get_attention(dummy)
print('attention shape:', attn.shape)  # (1, seq_length + 1)
````
